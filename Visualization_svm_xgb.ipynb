```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split 
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,classification_report,confusion_matrix
```

```python
df = pd.read_csv("yeni_egitim_verisi.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna() #null degerleri dropla
```

```python
df.isnull().sum()
```

```python
df.loc[df['label'] == 'phishing','label'] = 2
df.loc[df['label'] == 'spam','label'] = 1
df.loc[df['label'] == 'ham','label'] = 0
```

```python
df['text'] = df['subject'] + df['clean_body'] 
```

```python
df
```

```python
X = df['text']
y = df['label']
```

```python
X_train, X_test,y_train, y_test = train_test_split(X,y, test_size= 0.3, random_state= 42)
```

```python
X_train = X_train.fillna('')
X_test = X_test.fillna('')
y_train = y_train.fillna('')
y_test = y_test.fillna('')
```

```python
X # controlleri saglayalim
```

```python
y
```

```python
feature_extraction = TfidfVectorizer(lowercase= True, ngram_range=(1,2))
```

```python
X_train_features = feature_extraction.fit_transform(X_train)
X_test_features = feature_extraction.transform(X_test)
```

```python
svm = SVC()
svm.fit(X_train_features, y_train)

svm_train = svm.predict(X_train_features)
svm_test = svm.predict(X_test_features)


svm_train_acc = accuracy_score(y_train,svm_train)
svm_test_acc = accuracy_score(y_test, svm_test)

svmpre = precision_score(y_test,svm_test, average='weighted')
svmrec = recall_score(y_test,svm_test,average='weighted')
svmf1 = f1_score(y_test,svm_test,average='weighted')
```

```python
print("SVM metrikleri")
print("\nTrain Accuracy score:" ,svm_train_acc)
print("\nTest Accuracy score:", svm_test_acc)
print("\nPrecision Score:", svmpre)
print("\nRecall Score:", svmrec)
print("\nF1 Score:", svmf1)
print("\n Classification Report :", classification_report(y_test,svm_test)) 
print("\n Confusion matrix :",confusion_matrix(y_test,svm_test)) 


```
